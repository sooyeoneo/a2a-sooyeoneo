#### Text Splitter (청크 분할)

* 긴 회사 소개나 설명, 재무 이력 등은 토큰 최적화를 위해 적절한 크기로 분할
* LangChain의 `RecursiveCharacterTextSplitter` 등 사용

* 분할 목적 : 불필요한 토큰 낭비를 막고, 관련 정보만 LLM에 전달

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

def split_documents(docs, chunk_size=1000, chunk_overlap=50):
    splitter = RecursiveCharacterTextSplitter(chunk_size=chunk_size, chunk_overlap=chunk_overlap)
    return splitter.split_documents(docs)g
```

#### LangSmith 추적 연동 설정 코드

* 이 셀은 실행만 하고, 내용을 수정하거나 해석할 필요가 없음

* Jupyter Notebook에서 출력 숨기기
    * `%%capture` 매직커맨드 사용하면 실행 결과만 숨김 
    (코드는 노출 : 문서화 할 땐 nbconvert로 markdown/pdf로 export 후 해당 셀 삭제하면 코드 안 보임)    

```python
import os
from dotenv import load_dotenv

def init_env_and_langsmith(project_name="LEADSCOUT", tracing=True):
    load_dotenv()
    os.environ["LANGCHAIN_TRACING_V2"] = "true" if tracing else "false"
    os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGSMITH_API_KEY")
    os.environ["LANGCHAIN_PROJECT"] = "LEADSCOUT"

def print_langsmith_status():
    tracing = os.getenv("LANGCHAIN_TRACING_V2", "").lower()
    if tracing == "true":
        print("LangSmith 추적을 시작합니다.")
    else:
        print("LangSmith 추적을 하지 않습니다.")

def handle_langsmith_env_error():
    if not os.getenv("LANGCHAIN_API_KEY"):
        raise EnvironmentError("LANGSMITH_API_KEY가 .env에 없거나 잘못되었습니다.")  
```

#### LangSmith & 모니터링
* `.env` 설정으로 LangSmith Trance 활성화

* 토큰, 비용, 프롬프트, 응답 품질 등 실시간 대시보드 제공 (LangChain 공식 문서 및 LangSmith 대시보드 참고)

```python
init_env_and_langsmith(project_name="LEADSCOUT", tracing=True)
handle_langsmith_env_error()
print_langsmith_status()

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o")
result = llm.invoke("삼성전자 소개해줘")
print(result)
```

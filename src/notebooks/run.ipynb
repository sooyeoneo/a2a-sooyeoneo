### 🖥️ 실제 실행

* `python src/run_enhanced_rag.py`

### ⚙️ 실행 명령어

* 기본 실행
    * `python src/run_enhanced_rag.py`

* 특정 회사 분석
    * `python src/run_enhanced_rag.py --company "삼성전자"`
    
* 특정 필드만 검색
    * `python src/run_enhanced_rag.py --fields 대표자 매출액 직원수`

* 다른 데이터 파일 사용
    * `python src/run_enhanced_rag.py --data-file "path/to/your/company_data.json"`

* OpenAI API 키 지정
    * `python src/run_enhanced_rag.py --openai-key "your-openai-api-key"`

* 결과 저장 위치 변경
    * `python src/run_enhanced_rag.py --output "my_reports"`

* 모든 옵션 조합
    * `python src/run_enhanced_rag.py \`
        `--company "삼성전자" \`
        `--fields 대표자 매출액 직원수 설립년도 \`
        `--output "custom_reports" \`
        `--openai-key "your-key"`

* 도움말 보기
    * `python src/run_enhanced_rag.py --help`

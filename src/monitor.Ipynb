#### LangSmith & 모니터링
* `.env` 설정으로 LangSmith Trance 활성화

* 토큰, 비용, 프롬프트, 응답 품질 등 실시간 대시보드 제공 (LangChain 공식 문서 및 LangSmith 대시보드 참고)

```python
import os

# .env 파일 자동 로드
from dotenv import load_dotenv
load_dotenv()

# LangSmith 트레이싱 활성화 (환경변수 기반)
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGSMITH_API_KEY")
os.environ["LANGCHAIN_PROJECT"] = "LEADSCOUT"

# LangSmith 추적 허용하지 않을 때는 "LANGCHAIN_TRACING_V2" = "false"
```

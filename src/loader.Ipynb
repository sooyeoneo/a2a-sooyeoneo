#### Document Loader (기업 정보 수집 및  로드)
* Document Loader : 외부 데이터(웹, DB, API 등)을 파싱하여 Python 내 문서 객체로 변환하는 작업

* 크롤러, API, 또는 수동 데이터 -> 파싱 후 Python dict/list/JSON 형태로 가공
* 크롤링 또는 수집한 회사 JSON 리스트 -> 문서화

```python
def load_company_documents(file_path: str):
    import json
    with open(file_path, "r") as f:
        data = json.load(f)
    return data 
```

JSON 파싱 및 임시 저장

```python
import json

def save_to_json(data, file_path):
    """데이터(data)를 file_path 경로에 JSON 파일로 저장"""
    with open(file_path, "w") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

def load_form_json(File_path):
    with open(file_path):
        with open(file_path, "r") as f:
            return json.load(f)
```

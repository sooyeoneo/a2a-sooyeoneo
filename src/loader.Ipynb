#### Document Loader (기업 정보 수집 및  로드)
* Document Loader : 외부 데이터(웹, DB, API 등)을 파싱하여 Python 내 문서 객체로 변환하는 작업

* 크롤러, API, 또는 수동 데이터 -> 파싱 후 Python dict/list/JSON 형태로 가공
* 크롤링 또는 수집한 회사 JSON 리스트 -> 문서화

```python
def load_company_documents(file_path: str):
    import json
    with open(file_path, "r") as f:
        data = json.load(f)
    return data 
```

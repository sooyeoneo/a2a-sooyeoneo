##### 💭 실행하기 전에

* 터미널 현재 경로를 src/run으로 이동해야 함.
* `cd src/run`

* 현재 위치를 확인 후 `ls` 
* 아래 명령어 실행
    * `python run_web_crawling_rag.py "회사명 입력" --verbos`

```python
# 기본 사용법 (예시)
python run_web_crawling_rag.py "삼성전자"

# 사용자 지정 출력 디렉토리 (예시)
python run_web_crawling_rag.py "네이버" --output-dir "my_reports"

# 상세 로그와 함께 실행 (예시)
python run_web_crawling_rag.py "카카오" --verbose
```
